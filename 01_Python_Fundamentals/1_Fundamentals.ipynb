**Python lists** are built-in, mutable data structures used to store ordered collections of items inside square brackets [].

**Ordered** means the collection remembers the sequence in which items were added.
**Indexed** means you can access items directly using a number representing their position.

**Essential Properties**

1. Ordered: Items maintain their specific insertion sequence.

2. Mutable: You can modify, add, or remove items after creation.

3. Dynamic: They automatically resize as elements are added or removed.

4. Heterogeneous: A single list can hold strings, integers, booleans, and other lists.

5. Variable Size: Memory allocated dynamically grows or shrinks as elements change

6. Unhashable: Cannot be used as dictionary keys or stored inside sets because their values can change

```python
# Lists (Arrays in C, slighlty diff from arrays)
# Lists
print([1,5,6])
print([1,"priya",True, 7.8])
```

**Tuple** in Python is a built-in data structure used to store an ordered, unchangeable (immutable) collection of elements inside round brackets ().

**Essential Properties**

1. Immutability: Read-only after creation.

2. Duplicates: Allowed.

3. Data Types: Can hold mixed (heterogeneous) data types simultaneously.

4. Hashable: Can be used as dictionary keys or set elements, provided all items inside the tuple are also immutable.

5. Fixed Size: Memory size is locked at creation, making them highly memory-efficient.

```python
print((2,4,5.6,8,True))
```

**Set** in python is a built-in data structure used to store an unordered collection of unique elements and are mutable  

Unlike lists or tuples, sets automatically eliminate duplicates and do not track the position of items.   
**Essential Properties**  
1. Syntax: Written with curly brackets {} or created using the set() constructor.   2. Unordered: Items have no fixed sequence; they appear in random order each time you print them.  
3. No Duplicates: Every element inside a set must be entirely unique.  
3. Unindexed: You cannot access elements using numbers (e.g., my_set[0] will crash)

Applications
1. Removing Duplicates Instantly : If you have a collection of data with repeating items and you only want the unique values, converting it to a set is the cleanest and fastest way to clean it up.
2. High-Speed "Membership Testing" : If you need to check if an item exists inside a collection, searching a list is slow, but searching a set is near-instant.To find an item in a list, Python must check every single item one by one from start to finish.To find an item in a set, Python uses a shortcut called hashing to look exactly where the item should be. Even if your set has 10 million items, the check happens instantly
3. Comparing Data Collections (Venn Diagrams) : Sets excel at finding commonalities or differences between two groups of data. Doing this with lists requires writing complex nested loops, but sets handle it in one line.Social Media (Mutual Friends): Finding users who follow both you and a celebrity (Intersection).E-commerce Product Filtering: Finding items that have all selected tags like "Waterproof" AND "Wireless".
4. Tracking : "Seen" Items in AlgorithmsWhen writing code that explores networks, scrapes websites, or processes complex streams of data, you need to remember where you have already been so you do not get stuck in an infinite loop. Sets are the industry standard tool for tracking "visited" items.

```python
# Sets
print({1,4,True,4,2.5})

# Sets automatically remove duplicate values
'''In Python, the boolean True is a subclass of integers and shares the exact
 same value and hash as the integer 1 (True == 1 evaluates to True).
 Because the integer 1 was listed first, Python views True as a duplicate
 and drops it. '''
```

**Dictionary** in Python is a built-in data structure used to store data in key-value pairs.
It acts like a real-world dictionary or phonebook, where you look up a unique word (the key) to find its definition (the value).   

**Essential Properties:**
1. Syntax: Written with curly brackets and colons: {"key": "value"}.
2. Ordered: dictionaries maintain the exact insertion order of keys.
3. Unique Keys: Keys must be completely unique. If you add a duplicate key, it overwrites the old value.
4. Mutable: You can add, change, or remove key-value pairs at any time.
5. Not Indexed by Number: You cannot look up items using positional numbers like dict[0]. You must use the actual key name


**Strict Key Constraints**
While values can be absolutely anything (lists, sets, other dictionaries), keys face two strict rules:
1. Keys must be unique: You cannot have two identical keys.
2. Keys must be immutable: Keys must use unchangeable data types like strings, integers, or tuples. You cannot use a list or a set as a dictionary key.

```python
# Dictionary
print({'key' : 'value', 'name' :'priya', 'age' : 25, 'status':'UNEMPLOYED'})
```

**Dynamic typing** in Python means variable types are determined at runtime, not during compilation. You do not need to declare the data type of a variable when you create it, and the same variable can point to values of different types over its execution.  
EX : Python, JS, Ruby  

**Static typing** means variable types are checked and enforced at compile time, before the program ever runs. You must explicitly declare a variable's data type (or let the compiler infer it), and a variable cannot hold a value of a different type during execution.

Languages like Java, C++, Rust, Go, and TypeScript use static typing.

In **statically typed languages** like C++ or Java, variables are containers with fixed types. In Python, variables are simply names (labels) attached to objects stored in memory.


```python
# Dynamic Typing
name = 'seetha'

# Static Typing
int age = 23
```

**Binding**  
In Python, binding simply means attaching a name to an object in memory.

Unlike languages like C++ or Java where a variable is a physical container holding data, in Python, variables are just named tags or labels stuck onto objects.

**What is Name Binding?**
When you write x = 10, Python does two things:

1. Creates an integer object 10 in memory.

2. Binds the name x to that object.

If you reassign x = "hello", you aren't changing the 10 in memory. You are just unbinding x from 10 and binding it to a new string object "hello".

**Dynamic Binding** in Python (Late Binding)
Because Python evaluates code line-by-line at runtime, function and method lookup happens dynamically when the code is executed, not when it is compiled.

Example: Function Lookup at Runtime
Python doesn't lock in what a function name refers to when you define it; it resolves the name when the function is actually called:



Name Binding: Assigning a variable name to point to a memory location (name = object).



```python
# Dynamic Binding
a = 'rama'
print(a)
a = 'sita'
print(a)
```

```python
a = 'rama'
a = 'sita'
print(a)

# If you reassign a = "sita", you aren't changing the rama in memory.
# You are just unbinding x from rama and binding it to a new string object "sita"
```

**Keywords** (also called reserved words) are special words built directly into the programming language that have predefined meanings and specific instructions attached to them.

1. Pre-assigned meaning: The language engine already knows what a keyword is supposed to do (e.g., control flow, define functions, perform logical operations).

2. Cannot be used as names: You cannot name your variables, functions, or classes after a keyword.

**Identifiers** are the custom names you create to identify entities in your code, such as variables, functions, classes, arrays, or modules.

1. User-created: You choose what to call them (age, total_price, calculate_sum).

2. Name binding: An identifier acts as a label bound to a value, object, or block of code in memory.

**Standard Rules for Naming Identifiers**

1. Allowed Characters: Letters (a-z, A-Z), digits (0-9), and underscores (_).

2. Cannot Start with a Number: user1 is valid, but 1user is invalid.

3. Cannot Be a Keyword: You cannot name a variable if = 5 or class = "A".

4. Case-Sensitive: age, Age, and AGE are treated as three completely different identifiers in languages like Python, Java, C++, and JavaScript.

5. No Special Symbols or Spaces: Characters like @, $, #, -, or spaces are not allowed inside identifier names.

**Type conversion**  
Never makes changes in original data, instead creates new value in the memory and we work/operate on that new data.

```python
# Implicit Type Conversion - done by interpreter
a = 10
b = 4.8
c = a+b
print(c)
print('a',type(a))
print('b',type(b))
print('c',type(c))
print('''Implicit fails when there are diff data types which
dont support implicit type conversion, ex string and int''')
```

```python
# Literals - raw value stored in the variable or value of variable
# a(varibale) =(operator) 2(literal)

# Binary Literal(Base 2) - They use digits from 0 to 1
# Prefix the number with 0b or 0B
B = 0b1010
print(B)

# Decimal Literal(Base 10) - They use digits from 0 to 9
# Cannot begin with a leading zero
D = 10
print(D)

#  Octal Literal(Base 8) - They use digits from 0 to 7
# Prefix the number with 0o or 0O
O = 0o123 # 8^3 * 1 + 8^2 * 2 + 8^1 *3
print(O)

# Convert octal string to base 10 integer
decimal_val1 = int("12", 8)

# Hexa decimal numbers(Base 16)
# Prefix the number with 0x or 0X
# uses positional system based on powers of 16, moving from right to left
'''They use digits from 0 to 19 to represent values zero to nine,
and letters A to F to represent values ten to fifteen.'''

H = 0x12c
print(H)

print(bin(10))
print(oct(10))
print(hex(10))
```

```python
#Float Literals
f1 = 10
f2 = 1.5e2      # 1.5 * 10^2
f3 = 1.5e-3     # 1.5 * 10^-3
print(f1,f2,f3, sep = '\n')
```
